Lets start modeling our first neural network.

Steps:
1. Creating a model: We need to piece together the layer of neural network and build a model using sequential api.
2. Compiling a model: defining

```python
#
tf.random.set_seed(42)
```

```python
# Creating a Model: We need to piece together the layer of neural network and
# build a model using sequential api.

model_1 = tf.keras.Sequential([
    tf.keras.layers.Dense(1)
])

# Compiling a model

model_1.compile(loss = tf.keras.losses.BinaryCrossentropy(),
              optimizer = tf.keras.optimizers.SGD(),
              metrics = ['accuracy'])

# Fitting a model

model_1.fit(X, y, epochs = 5)
```

```python
# Train the model for longer

model_1.fit(X, y, epochs = 50)
```

```python
model_1.fit(X, y, epochs = 200)
```

### Not good result of module so some chnage in hidden layer so model_2

```python
# Creating a Model: We need to piece together the layer of neural network and
# build a model using sequential api.

model_2 = tf.keras.Sequential([
    tf.keras.layers.Dense(1),
    tf.keras.layers.Dense(1),
    tf.keras.layers.Dense(1)
])

# Compiling a model

model_2.compile(loss = tf.keras.losses.BinaryCrossentropy(),
              optimizer = tf.keras.optimizers.SGD(),
              metrics = ['accuracy'])

# Fitting a model

model_2.fit(X, y, epochs = 5)
```

```python
# Evaluate the model
model_2.evaluate(X, y)
```
