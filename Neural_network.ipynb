```python
from sklearn.datasets import make_circles
```

```python
# lets create 1000 samples

n_samples = 1000

# create a circle

X, y = make_circles(n_samples,
                  noise = 0.03,
                  random_state = 42)
```

```python
# lets check out the features
X
```

```python
# lets see the first 10 labels
y[:10]
```
