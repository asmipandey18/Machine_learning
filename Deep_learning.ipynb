# Introduction to Tensors





```python
# lets check if we have tensorflow installed in our system or not

import tensorflow as tf

#@ To check the version of tensorflow libary
tf.__version__
```

- Typically, you wont be creating tensors by yourself.
- Tensorflow has a built in module(tf.io/ tf.data) which are able to read your data sources and automatically convert it into corresponding tensors and later on the neural network models will process those tensors for model building

```python
# Lets create tensors with tf.constant()

# Create a scalar (rank 0 tensor)

scalar = tf.constant(7)
scalar
```

A scalar is known as rank 0 tensor, Because it has no dimension. (just a number)

```python
scalar.ndim
```

```python
#@ Create a vector (more than 0 dimension, i.e. 1st dimension)

vector = tf.constant([10,10])
vector
```

```python
vector.ndim
```

```python
#@ Create a matrix (more than 1 dimension i.e. 2)

matrix = tf.constant([[10, 7],
                      [7, 10]])
```

```python
matrix
```

```python
matrix.ndim
```
