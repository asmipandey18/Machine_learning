Random Tensors:

- Random tensors are tensors of some arbitrary size which contains some random numbers

why do we need such random tensor??

ANS: when neural network use to initialize the weight(patterns) then, they are trying to learn from the data.



````python
#@ Creating Random Tensors

random_1 = tf.random.Generator.from_seed(42) # set the seed for reproducibality
random_1 = random_1.normal(shape=(3, 2)) # create tensor from normal distribution
random_2 = tf.random.Generator.from_seed(42)
random_2 = random_2.normal(shape=(3, 2))

# are these equal?

random_1, random_2, random_1 == random_2
````

````python
#@ Creating Random Tensors

random_3 = tf.random.Generator.from_seed(42) # set the seed for reproducibality
random_3 = random_3.normal(shape=(3, 2)) # create tensor from normal distribution
random_4 = tf.random.Generator.from_seed(11)
random_4 = random_4.normal(shape=(3, 2))

# are these equal?

random_3, random_4, random_3 == random_4
````

````python
#@ Shuffle a tensor ( it is required when you want to shuffle your data)

not_shuffled = tf.constant([[10, 7],
                           [3, 4],
                           [2, 5]])

tf.random.shuffle(not_shuffled)
````

````python
tf.random.set_seed(42)

tf.random.shuffle(not_shuffled, seed=42)
````

## Matrix Multiplication

- The inner dimension must match for matrix multiplication; 1st column and 2nd row size must match

- the resulting matrix size would be 1st row 2nd column.

````python
# print(tensor)
tf.matmul(tensor, tensor)
````

````python
# python operator
tensor @ tensor
````

````python
# lets create (3, 2) tensor

X = tf.constant([[1, 2],
                 [3, 4],
                 [5, 6]])

# Create another (3, 2) tensors
Y = tf.constant([[7, 8],
                 [9, 10],
                 [11, 12]])

X, Y
````

````python
# try to matrix multiple
X @ Y
````

Error is due to the mismatch in inner dimension of two matrix,. To fix this, we need to

- Reshape X to (2, 3) so its (2, 3) @ (3, 2)
- Reshape Y to (3, 2) so its (3, 2) @ (2, 3)

we can achieve this using:

```tf.reshape()``` - allows you to reshape a tensor in a defined shape

```tf.transpose()``` - switches the dimension of a given tensor

````python
# Reshape (3, 2) -> (2, 3)
tf.reshape(Y, shape = (2, 3)
)
````

````python
X @ tf.reshape(Y, shape = (2, 3))
````

````python
# Lets try to transpose (3, 2) -> (2, 3)
tf.transpose(X)
````

````python
# if we try matmul
tf.matmul(tf.transpose(X), Y)
````

## Finding min, max, mean, sum in Tensors

- tf.reduce_max()
- tf.reduce_mean()
- tf.reduce_sum()

````python
tf.math.reduce_std(tf.cast(E, tf.float32))
````

````python
# tf.function
# Creating a simple function

def function(x, y):
    return x ** 2 + y

x = tf.constant(np.arange(0, 10))
y = tf.constant(np.arange(10, 20))

print(x)
print(y)

function(x, y)
````

````python
# Create a same function and decorate it tf.function
@tf.function

def tf_function(x, y):
    return x ** 2 + y

tf_function(x, y)
````
