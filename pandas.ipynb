```python
df.info()
```

```python
#@ which was the most ordered item
# groupby
i = df.groupby('item_name')
i = i.sum()
i = i.sort_values(['quantity'], ascending=False)
i.head(1)
```

```python
#@ what was the most oredered item in the choice descriptive coulumns
df.groupby('choice_description').sum().sort_values(['quantity'], ascending=False).head(1)
```
