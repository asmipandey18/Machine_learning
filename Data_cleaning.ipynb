```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("/content/my_file (1).csv")
```

```python
df["peak"] = df["peak"].fillna("delete_me")
```

```python
df["all_time_peak"] = df["all_time_peak"].fillna("do_not_delete_me")
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.strip()
df.columns = df.columns.str.replace(" ", "_")
df.columns = df.columns.str.replace(r"[^a-zA-Z0-9_]", "", regex=True)
print(df.columns)
```

```python
df.rename(columns={
    "actualgross": "actual_gross",
    "adjustedgross_in_2022_dollars": "adjusted_gross_in_2022_dollars"
}, inplace=True)
```

```python
print(df.columns)
```

```python
df["peak"] = df["peak"].str.replace(r"\[.*?\]", "", regex=True)
```

```python
print(df["peak"])
```

```python
df["all_time_peak"] = df["all_time_peak"].str.replace(
    r"\[.*?\]", "", regex=True
)
```

```python
print(df["all_time_peak"])
```

```python
df["actual_gross"] = df["actual_gross"].str.replace(r"\[.*?\]", "", regex=True)

df["actual_gross"] = df["actual_gross"].str.replace("$", "", regex=False)
df["actual_gross"] = df["actual_gross"].str.replace(",", "", regex=False)

df["actual_gross"] = pd.to_numeric(df["actual_gross"])
```

```python
print(df["actual_gross"].head())
print(df["actual_gross"].dtype)
```

```python
df["adjusted_gross_in_2022_dollars"] = df["adjusted_gross_in_2022_dollars"].str.replace("$", "", regex=False)

df["adjusted_gross_in_2022_dollars"] = df["adjusted_gross_in_2022_dollars"].str.replace(",", "", regex=False)

df["adjusted_gross_in_2022_dollars"] = pd.to_numeric(
    df["adjusted_gross_in_2022_dollars"]
)
```

```python
df["adjusted_gross_in_2022_dollars"].head()
```

```python
df["adjusted_gross_in_2022_dollars"].dtype
```

```python
df["tour_title"] = df["tour_title"].str.replace(
    r"[^a-zA-Z0-9\s]", "", regex=True
)
```

```python
df["tour_title"].head()
```

```python
def clean_year(x):
    x = str(x).replace("–", "-")

    if "-" in x:
        start, end = x.split("-", 1)

        if len(end) == 2:
            end = start[:2] + end

        return f"{start}, {end}"

    return x

df["years"] = df["years"].apply(clean_year)
```

```python
df["years"].head()
```

```python
df["average_gross"] = df["average_gross"].str.replace("$", "", regex=False)

df["average_gross"] = df["average_gross"].str.replace(",", "", regex=False)

df["average_gross"] = pd.to_numeric(df["average_gross"])
```

```python
print(df["average_gross"].to_string(index=False))
```

```python
print(df["average_gross"].dtype)
```

```python
df.drop("ref", axis=1, inplace=True)
```
