Check Number of Dimensions in an Numpy Array?

```python
## Check Number of dimension in an NumPy Array?
import numpy as np

a = np.array(42)
b = np.array([1, 2, 3,])
c = np.array([[1, 2, 3], [4, 5, 6]])
d = np.array([[[1,2,3], [2,3,4]], [[3,5,6,], [6,5,8]]])

print(a.ndim)
print(b.ndim)
print(c.ndim)
print(d.ndim)
```

```python
# create an array of 5 dimension and verify its dimension.

import numpy as np

arr = np.array([1,2,3,4], ndmin=10)

print(arr)
print('number of dimensions :', arr.ndim)


```
